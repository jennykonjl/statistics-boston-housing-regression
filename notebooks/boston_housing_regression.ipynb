#**Linear regression with Boston data example**
***

Here,

$Y$ = CRIM (also called "target" data in python)

and

$X$ = POLY(MEDV,2) (or independent variables)

which we will use to fit a linear regression model and predict CRIM. We will use the least squares method as the way to estimate the coefficients.  

We'll use two ways of fitting a linear regression using statsmodels or sklearn. We recommend the first but the second is also powerful in its features.

> **Fitting Linear Regression using `statsmodels`**
***

This formula notation will seem familiar to `R` users, but will take some getting used to for people coming from other languages or are new to statistics.
**bold text**
The formula gives instruction for a general structure for a regression call. For `statsmodels` (`ols` or `logit`) calls you need to have a Pandas dataframe with column names that you will add to your formula. In the below example you need a pandas data frame that includes the columns named (`Outcome`, `X1`,`X2`, ...), bbut you don't need to build a new dataframe for every regression. Use the same dataframe with all these things in it. The structure is very simple:

`Outcome ~ X1`

But of course we want to to be able to handle more complex models, for example multiple regression is doone like this:

`Outcome ~ X1 + X2 + X3`

This is the very basic structure but it should be enough to get you through the homework. Things can get much more complex, for a quick run-down of further uses see the `statsmodels` [help page](http://statsmodels.sourceforge.net/devel/example_formulas.html).

```python
import statsmodels.formula.api as smf

# List of predictor variables
predictors = ['MV', 'RM', 'PT', 'ZN', 'INDUS', 'CHAS', 'NOX', 'AGE', 'DIS', 'RAD', 'TAX', 'B', 'LSTAT']

for predictor in predictors:
    formula = f"CRIM ~ {predictor}"
    model = smf.ols(formula=formula, data=boston).fit()
    print(f"Simple Linear Regression: CRIM ~ {predictor}")
    print(model.summary())
    print("\n" + "=" * 80 + "\n")
```

```python
# Multiple Regression

MLRmodel = ols('CRIM ~ MV + RM + PT + ZN + INDUS + CHAS + NOX + AGE + DIS + RAD + TAX + B + LSTAT', boston).fit()
print(MLRmodel.summary())
```

### MLR (R-squared, P-value and Coefficient)

```python
# R-squared value
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print("R-squared:", model.rsquared)

# P-values for each predictor
p_values = MLRmodel.pvalues
print("\nP-values:")
print(p_values)

# Coefficients for each predictor (including intercept)
coefficients = MLRmodel.params
print("\nCoefficients:")
coefficients
```

> ## Scatter Plot

```python
# Plot: Simple vs. Multiple Regression Coefficients

# Get simple regression coefficients
simple_reg_coeffs = {}
for predictor in predictors:
        formula = f'CRIM ~ {predictor}'
        model = ols(formula, data=boston).fit()
        # Get the coefficient for the predictor variable (not the intercept)
        simple_reg_coeffs[predictor] = model.params[predictor]


# Get multiple regression coefficients
mlr_coefficients = MLRmodel.params.drop('Intercept')

# Create a scatter plot
plt.figure(figsize=(10, 8))
for predictor in simple_reg_coeffs.keys():
    if predictor in mlr_coefficients:
        plt.scatter(simple_reg_coeffs[predictor], mlr_coefficients[predictor], label=predictor)

plt.xlabel("Simple Regression Coefficient")
plt.ylabel("Multiple Regression Coefficient")
plt.title("Scatter Plot of Simple vs. Multiple Regression Coefficients for CRIM")
plt.legend()
plt.grid(True)
plt.show()
```

> ## Table

```python
# Table: Simple vs. Multiple Regression Coefficients

# Prepare data for table
coefficients_df = pd.DataFrame({
    'Predictor': simple_reg_coeffs.keys(),
    'Simple Regression Coeff': simple_reg_coeffs.values()
})

# Merge with multiple regression coefficients
mlr_coeffs_df = mlr_coefficients.reset_index().rename(columns={'index': 'Predictor', 0: 'Multiple Regression Coeff'})
coefficients_table = pd.merge(coefficients_df, mlr_coeffs_df, on='Predictor')

# Display the table
print("\nComparison of Simple vs. Multiple Regression Coefficients for CRIM:")
coefficients_table
```

# Polynomial Regression Analysis
***

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

df = boston

# Define the response variable
Y = df['CRIM']

# Create polynomial features
poly = PolynomialFeatures(degree=3)

# Initialize the plot
plt.figure(figsize=(20, 15))

# Loop through each predictor
for i, predictor in enumerate(boston.columns):
    # Define the predictor variable
    X = df[[predictor]]

    # Create polynomial features
    X_poly = poly.fit_transform(X)

    # Fit the model
    model = LinearRegression()
    model.fit(X_poly, Y)

    # Predict values
    Y_pred = model.predict(X_poly)

    # Create subplot
    plt.subplot(4, 4, i+1)
    plt.scatter(X, Y, color='blue', label='Data Points')
    plt.plot(X, Y_pred, color='red', label='Polynomial Fit')
    plt.xlabel(predictor)
    plt.ylabel('CRIM')
    plt.title(f'{predictor} vs CRIM')
    plt.legend()

# Adjust layout and show the plot
plt.tight_layout()
plt.show()
```

```python
# Define predictors and target variable
predictors = ['MV', 'RM', 'PT', 'ZN', 'INDUS', 'CHAS', 'NOX', 'AGE', 'DIS', 'RAD', 'TAX', 'B', 'LSTAT']
y = boston['CRIM']

# Store results for comparison
results = []

# **Step 1: Fit Polynomial Regression for Each Predictor**
for col in predictors:
    X = boston[col]

    # Create polynomial features (X, X², X³)
    X_poly = pd.DataFrame({'X': X, 'X2': X**2, 'X3': X**3})
    X_poly = sm.add_constant(X_poly)  # Add intercept

    # Fit regression model
    model = sm.OLS(y, X_poly).fit()

    # Store p-values for X² and X³ terms
    results.append({
        'Predictor': col,
        'P-value (X²)': model.pvalues['X2'],
        'P-value (X³)': model.pvalues['X3']
    })

# **Step 2: Convert Results to DataFrame**
results_df = pd.DataFrame(results)

# **Step 3: Display Table**
print(results_df)
```

```python

import matplotlib.pyplot as plt
# **Residual Errors**
# ***
# Calculate residuals
residuals = MLRmodel.resid

# Create residual plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x=MLRmodel.fittedvalues, y=residuals)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel("Fitted Values (Predicted CRIM)")
plt.ylabel("Residuals")
plt.title("Residual Plot of Multiple Linear Regression Model")
plt.show()

# Plot distribution of residuals (histogram or Q-Q plot)
plt.figure(figsize=(10, 6))
sns.histplot(residuals, kde=True)
plt.xlabel("Residuals")
plt.ylabel("Frequency")
plt.title("Distribution of Residuals")
plt.show()

# Optional: Q-Q plot for checking normality of residuals
sm.qqplot(residuals, line='s')
plt.title("Q-Q Plot of Residuals")
plt.show()
```
